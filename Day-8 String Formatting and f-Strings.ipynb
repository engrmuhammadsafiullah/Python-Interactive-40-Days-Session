```python
item = "Laptop"
quantity = 3
price_each = 899.99

# The legacy way (Concatenation)
print("Item: " + item + " | Total: " + str(quantity * price_each))

# The modern way (f-Strings)
print(f"Item: {item} | Total: ${quantity * price_each:.2f}") 
# Note: :.2f limits the decimal output to exactly 2 places.
```
