```python
# Integers (int) Whole numbers without decimals. 
#They can be positive, negative, or zero.
age = 25
temperature = -5
items = 0
#The type() Tool: Introduce the built-in function used to inspect any variable's data type.
print(type(age)) 

```

```python
#Floating-Point Numbers (float)
#Numbers containing a decimal point or scientific notation.
price = 19.99
pi = 3.14159
height = 1.8

print(type(price)) 
```

```python
#Strings (str)
#Text wrapped in single (' '), double (" "), or triple (""" """) quotes.
name = "Alice"
greeting = 'Hello, World!'
multiline = """This text spans
multiple lines."""

print(type(name))  

```

```python
#Booleans (bool)Binary logical states.
#They can only ever be True or False (capitalization matters).
is_logged_in = True
has_passed = False

print(type(is_logged_in))  
```
